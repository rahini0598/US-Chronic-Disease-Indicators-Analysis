# **Data Cleaning,Data Transformation and Feature Engineering**

```python
df.isnull().sum()
```

```python
#Remove Duplicate Rows

df.duplicated().sum()

df.drop_duplicates(inplace=True)
```

```python
# Drop the Unnecessary columns

df.drop(['Response', 'DataValueFootnote'], axis=1, inplace=True)
```

```python
#Remove Missing DataValue Rows
df = df.dropna(subset=['DataValue'])
```

```python
#Confidence Limits


df['LowConfidenceLimit'] = df['LowConfidenceLimit'].fillna(df['DataValue'])

df['HighConfidenceLimit'] = df['HighConfidenceLimit'].fillna(df['DataValue'])
```

```python
#Geolocation

df['Geolocation'].isnull().sum()
```

```python
numeric_cols = ['DataValue', 'DataValueAlt', 'LowConfidenceLimit', 'HighConfidenceLimit']

for col in numeric_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
df['Geolocation'].fillna(df['Geolocation'].mode()[0], inplace=True)
```

```python
#Remove Extra Spaces

obj_cols = df.select_dtypes(include='object').columns

df[obj_cols] = df[obj_cols].apply(lambda x: x.str.strip())
```

```python
print(df.isnull().sum())
```

```python
##Data Transformation
# Display the data type of each column

print(df.dtypes)
```

```python
#Convert Numeric Columns

numeric_cols = ['DataValue','DataValueAlt',
                'LowConfidenceLimit','HighConfidenceLimit']

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
#Convert Year Columns

df['YearStart'] = df['YearStart'].astype(int)

df['YearEnd'] = df['YearEnd'].astype(int)
```

```python
#Convert ID Columns into Category
cat_cols = [
'LocationAbbr',
'TopicID',
'QuestionID',
'DataValueTypeID',
'StratificationCategoryID1',
'StratificationID1'
]

for col in cat_cols:
    df[col] = df[col].astype('category')
```

```python
#Standardize Text Columns

text_cols = [
'LocationDesc',
'Topic',
'Question',
'StratificationCategory1',
'Stratification1'
]

for col in text_cols:
    df[col] = df[col].str.strip().str.title()
```

```python
#Extract Latitude & Longitude

df[['Longitude','Latitude']] = df['Geolocation'].str.extract(
r'POINT \(([-\d\.]+) ([-\d\.]+)\)'
)

df['Latitude']=pd.to_numeric(df['Latitude'])

df['Longitude']=pd.to_numeric(df['Longitude'])
```

```python
#sort dataset


df = df.sort_values(
['YearStart','LocationDesc']
)
```

```python
#Reset Index

df.reset_index(drop=True,inplace=True)
```

```python
#Rename Important Columns

df.rename(columns={
'LocationDesc':'State',
'DataValue':'DiseaseValue',
'Topic':'Disease'
}, inplace=True)
```

```python
df.info()
```

**Feature Engineering**

```python
#Region Type

import numpy as np

df['RegionType'] = np.where(
    df['State'] == 'United States',
    'National',
    'State/Territory'
)
```

```python
#Survey Type

survey_map = {
    'BRFSS': 'Behavioral Risk Survey',
    'NVSS': 'Vital Statistics',
    'YRBSS': 'Youth Risk Survey',
    'ACS': 'American Community Survey'
}

df['SurveyType'] = df['DataSource'].map(survey_map)
```

```python
#Confidence Width

df['ConfidenceWidth'] = (
    df['HighConfidenceLimit'] -
    df['LowConfidenceLimit']
)
```



# **Statistical analysis & Visualizations**


```python
#Correlation Analysis

numeric_df = df.select_dtypes(include='number')

print(numeric_df.corr())
```

```python
prevalence_df = df[df['DataValueType'] == 'Crude Prevalence']

col = 'DiseaseValue'

mean_val = prevalence_df[col].mean()
variance_val = prevalence_df[col].var()
std_val = prevalence_df[col].std()
cv_val = (std_val / mean_val) * 100

print("=== Measures of Dispersion ===")
print(f"Mean                 : {mean_val:.2f}")
print(f"Variance             : {variance_val:.2f}")
print(f"Standard Deviation   : {std_val:.2f}")
print(f"Coefficient of Variation : {cv_val:.2f}%")
```

```python
#skewness and kurtosis

prevalence_df = df[df['DataValueType'] == 'Crude Prevalence']

col = 'DiseaseValue'

skewness_val = prevalence_df[col].skew()
kurtosis_val = prevalence_df[col].kurt()

print("=== Skewness & Kurtosis Analysis ===")
print(f"Column     : {col}")

print(f"Skewness   : {skewness_val:.2f} "
      f"({'Right-skewed' if skewness_val > 0 else 'Left-skewed' if skewness_val < 0 else 'Symmetric'})")

print(f"Kurtosis   : {kurtosis_val:.2f} "
      f"({'Leptokurtic (Heavy Tails)' if kurtosis_val > 0 else 'Platykurtic (Light Tails)' if kurtosis_val < 0 else 'Mesokurtic (Normal Distribution)'})")

```

```python
# Box Plot (Outliers & Kurtosis)

prevalence_df = df[df['DataValueType'] == 'Crude Prevalence']

plt.figure(figsize=(7,5))

sns.boxplot(y=prevalence_df['DiseaseValue'], color='lightgreen')

plt.title("Box Plot of Disease Values (Crude Prevalence)")
plt.ylabel("Disease Value (%)")

plt.show()
```

#**Interpretation**


The box plot of Disease Values (Crude Prevalence) shows the distribution of disease prevalence percentages across the dataset. The median line represents the central tendency, while the box indicates the interquartile range (IQR), where the middle 50% of observations are concentrated. The whiskers represent the overall spread of the data, and any points beyond the whiskers indicate potential outliers. These outliers may reflect states, diseases, or demographic groups with unusually high or low prevalence rates. Overall, the box plot provides a clear understanding of data variability, dispersion, and the presence of extreme values in chronic disease prevalence.

#**Bivariate Analysis**

**Line Plot – Average Disease Indicator Value by Year**

---



```python
import matplotlib.pyplot as plt

# Convert YearStart to integer
df['YearStart'] = df['YearStart'].astype(int)

# Calculate average DiseaseValue by year
year_avg = df.groupby('YearStart')['DiseaseValue'].mean().reset_index()

plt.figure(figsize=(8,5))

plt.plot(
    year_avg['YearStart'],
    year_avg['DiseaseValue'],
    marker='o',
    linewidth=2,
    color='green'
)

plt.title("Line Plot – Average Disease Indicator Value by Year")
plt.xlabel("Year")
plt.ylabel("Average Disease Indicator Value")

plt.xticks(year_avg['YearStart'])

plt.grid(True)

plt.show()
```

**Interpretation :**

1. The line plot shows the trend in the average disease indicator values from 2020 to 2023, enabling a year-wise comparison of disease-related measurements.

2. The highest average disease indicator value was recorded in 2020 (819.23), indicating that disease indicator values were at their peak during this year.

3. The average value decreased to 675.95 in 2021, followed by an increase to 804.91 in 2022, showing fluctuations rather than a consistent upward or downward trend.

4. In 2023, the average disease indicator value dropped sharply to 26.45, which is significantly lower than the previous years. This may indicate incomplete data collection, fewer available records, or differences in the reporting period for 2023.

5. Across the four years, the overall average disease indicator value is 581.64, reflecting the general level of disease indicator values in the dataset.

6. Overall, the line plot indicates that disease indicator values varied considerably between years, with 2020 recording the highest average and 2023 showing a substantial decline that should be interpreted carefully in further analysis.
